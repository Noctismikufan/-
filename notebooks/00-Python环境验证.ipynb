## 3. 一段真实的数据处理

用 pandas 造一张成绩表，算平均分并排序 —— 这就是后续课程里最常见的数据分析套路。

```python
df = pd.DataFrame(
    {
        "姓名": ["小明", "小红", "小刚", "小美"],
        "数学": [88, 95, 72, 90],
        "英语": [76, 91, 85, 89],
        "计算机": [92, 87, 78, 96],
    }
)

df["总分"] = df[["数学", "英语", "计算机"]].sum(axis=1)
df["平均分"] = df[["数学", "英语", "计算机"]].mean(axis=1).round(1)
df = df.sort_values("总分", ascending=False).reset_index(drop=True)

print(df)
print("\n班级平均分:", df["平均分"].mean().round(2))
```

## 4. 画一张中文图表

Windows 下 matplotlib 默认字体不含中文，需要指定中文字体，否则中文会变成方块。

下面这两行是**固定写法**，建议背下来，以后每个画图脚本都加上。

```python
import matplotlib.pyplot as plt

# 解决中文显示为方块的问题（固定写法）
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False  # 解决负号显示异常

fig, ax = plt.subplots(figsize=(7, 4))
subjects = ["数学", "英语", "计算机"]
x = np.arange(len(subjects))
width = 0.2

for i, (_, row) in enumerate(df.iterrows()):
    offset = (i - (len(df) - 1) / 2) * width
    ax.bar(x + offset, row[subjects].values, width, label=row["姓名"])

ax.set_title("各科成绩对比")
ax.set_xticks(x)
ax.set_xticklabels(subjects)
ax.set_ylabel("分数")
ax.legend()
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. 跑一个机器学习小模型

用经典鸢尾花数据集训练一个分类模型，感受一下机器学习的基本流程：

**准备数据 → 切分训练/测试集 → 训练 → 评估**

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print(f"测试集准确率: {accuracy_score(y_test, pred):.2%}")
```
